# B± → π±π±π∓ — QMI em bins com x, y, dx, dy

Cópia do notebook 26. Cada um dos 25 bins constantes tem quatro parâmetros
compartilhados entre cargas, `S_QMI.bin_NN.x/y/dx/dy`:
`QMI_plus = (x+dx) + i(y+dy)` e `QMI_minus = (x-dx) + i(y-dy)`.
São parâmetros de dinâmica, pois alteram a forma da onda-S.

O coeficiente CPV global `S_QMI.x/y/dx/dy` continua livre e multiplica essas
formas. No primeiro bin, `x=1, y=dx=dy=0` ficam fixos. No bin do veto D0,
os quatro valores ficam fixos em zero. Os demais bins têm os quatro livres.
Mantemos 117 parâmetros livres, nenhuma normalização individual e nenhuma
interpolação ou penalidade de suavidade. A transformação das sementes preserva
as amplitudes iniciais do notebook 26 em ambas as cargas.

Os gráficos e comparações resolvem x±dx e y±dy antes de multiplicar pelo
coeficiente global. O helper local `CPBinPart` usa a resolução de parâmetros
da API; exportar esse modelo em JSON exige registrar esse helper como plugin.

## Dados

Os invariantes são lidos como `s12=m12Sq`, `s13=m13Sq`, `s23=m23Sq`, com a carga ±1.
O ROOT atual já está dobrado (`s13 <= s23`): isso é compatível com a amplitude simétrica,
mas exige dobrar as expectativas nos diagnósticos. O filtro por carga é explícito.

```python
def _load_charge_sample(charge_value):
    raw = read_root_tree(
        DATA_FILE, DATA_TREE,
        {"s12": "m12Sq", "s13": "m13Sq", "s23": "m23Sq", "charge": "charge"},
    )
    mask = np.asarray(raw["charge"]) == charge_value
    n = int(mask.sum())
    sample = PhaseSpaceSample(
        s12=jnp.asarray(raw["s12"])[mask],
        s13=jnp.asarray(raw["s13"])[mask],
        s23=jnp.asarray(raw["s23"])[mask],
        weights=jnp.ones(n, dtype=jnp.float64),
    )
    return D0_VETO.apply(sample)


plus_data = _load_charge_sample(+1)
minus_data = _load_charge_sample(-1)

print(f"B+ events (post D0 veto): {plus_data.size:,}")
print(f"B- events (post D0 veto): {minus_data.size:,}")
print(f"total: {plus_data.size + minus_data.size:,}  (fitTree has 95110 entries, 0 in the D0 window)")

for label, sample in (("B+", plus_data), ("B-", minus_data)):
    print(label, "fraction s13 <= s23:", float(jnp.mean(sample.s13 <= sample.s23)))
if plus_data.size + minus_data.size != int(N_TOTAL_EVENTS):
    raise ValueError("Observed event count differs from the input yield configuration")
```

## Eficiência (ACC)

Histogramas `Hacc_pipipi_15161718_Plus`/`Minus` em `(m', theta')`, `theta' in [0, 0.5]`
(`useUpperHalf=kTRUE` no Laura++) $\Rightarrow$ `folded=True`. `SquareDalitzHistogramEfficiency`
troca a busca por bin mais próximo pela interpolação bilinear do
Laura++ (`Lau2DHistDP::interpolateXY`); a eficiência é uma razão adimensional, então não leva
correção de jacobiano.

```python
def _load_square_dalitz_histogram2d(file_path, histogram):
    values, mp_edges, tp_edges = read_root_histogram2d(file_path, histogram)
    return jnp.asarray(values), jnp.asarray(mp_edges), jnp.asarray(tp_edges)


def _laura_efficiency(file_path, histogram):
    values, mp_edges, tp_edges = _load_square_dalitz_histogram2d(file_path, histogram)
    return SquareDalitzHistogramEfficiency(
        mprime_edges=mp_edges, thetaprime_edges=tp_edges, values=values,
        mother_mass=MOTHER_MASS, masses=DAUGHTER_MASSES, pair=SQDP_PAIR, folded=True, interpolation="spline", clip=True,
    )


plus_efficiency = _laura_efficiency(ACC_FILE, ACC_HIST_PLUS)
minus_efficiency = _laura_efficiency(ACC_FILE, ACC_HIST_MINUS)
print("ACC B+ grid:", plus_efficiency.values.shape, "  ACC B- grid:", minus_efficiency.values.shape)
```

## Background `qqbar` e `kpipibkg`

`SquareDalitzHistogramBackground` aplica a mesma interpolação bilinear e, além
disso, divide pelo jacobiano `|d(s13,s23)/d(m',theta')|` -- o histograma armazena uma densidade em
`(m', theta')` (`Lau2DHistDPPdf`), mas a amostra de normalização/a verossimilhança trabalham em
`(s13, s23)`. As duas ressonâncias de background do `.cc` são carregadas aqui; `kpipibkg` usa o
mesmo histograma `hbkg` para B+ e B- (como no `.cc`). O escalonamento pela assimetria de yield fixa
de cada uma (`qqbarEventsAsym=-0.0078`, `kpipibkgEventsAsym=0.03`) é feito mais abaixo, depois que
`plus_model`/`minus_model` existem -- o método `with_charge_asymmetry` normaliza cada forma pela sua
integral pós-veto *na mesma amostra de normalização/quadratura usada pelo ajuste* antes de aplicar
`(1 ∓ asym)`, reproduzindo exatamente `bkgndEvents = N_bkg * 0.5 * (1 - carga*asym)` do
`LauCPFitModel.cc`.

Na avaliação do PDF, `divide_jacobian=True` faz a conversão para a densidade em
`(s13,s23)`. Para geração, os geradores usam o método `generation_value()` da
classe, que retorna o valor bruto interpolado sem essa divisão, seguindo
`LauBkgndDPModel::generate` do Laura++.

```python
def _laura_background(file_path, histogram):
    values, mp_edges, tp_edges = _load_square_dalitz_histogram2d(file_path, histogram)
    return SquareDalitzHistogramBackground(
        mprime_edges=mp_edges, thetaprime_edges=tp_edges, values=values,
        mother_mass=MOTHER_MASS, masses=DAUGHTER_MASSES, pair=SQDP_PAIR, folded=True, interpolation="spline",  divide_jacobian=True,
    )


_qqbar_plus_pdf = _laura_background(QQBAR_BKG_FILE, QQBAR_HIST_PLUS)
_qqbar_minus_pdf = _laura_background(QQBAR_BKG_FILE, QQBAR_HIST_MINUS)
_kpipibkg_pdf = _laura_background(KPIPI_BKG_FILE, KPIPI_HIST)

print("qqbar B+ grid:", _qqbar_plus_pdf.values.shape, "  qqbar B- grid:", _qqbar_minus_pdf.values.shape)
print("kpipibkg grid:", _kpipibkg_pdf.values.shape)
```

## Fit isobárico nos dados → semente QMI

Primeiro ajustamos os nove isóbaros aos mesmos dados, eficiência, veto e backgrounds
utilizados pelo QMI. Nenhum componente é normalizado individualmente.
Após um fit válido, a soma bruta `sigma0 + f0_980 + f0_1370` é avaliada nos nós,
separadamente por carga, com os coeficientes CP ajustados. `chi_c0` permanece separado.
Os seis isóbaros restantes também começam dos parâmetros ajustados. A interpolação
da QMI aproxima essa soma entre os nós; não há reescalonamento da semente.


```python
SWAVE_NAMES = ("sigma0", "f0_980", "f0_1370")
isobar_shared = make_shared_coefficients()
isobar_dynamics = make_dynamics_parameters()


def build_isobar_components(charge):
    return [
        Resonance(
            name, RESONANCE_PAIR, isobar_shared[name].for_charge(charge),
            lineshape=LINESHAPES[name](), angular=ZemachP(),
            normalize_component=False,
            mass=isobar_dynamics[f"{name}.mass"], width=isobar_dynamics[f"{name}.width"],
            spin=SPINS[name], resonance_radius=4.0, parent_radius=4.0,
        )
        for name in LINESHAPES
    ]


isobar_models = {
    charge: DecayModel(channel, build_isobar_components(charge),
                       normalize_components=False, **NORMALIZATION_CONFIG)
    for charge, channel in ((+1, PLUS_CHANNEL), (-1, MINUS_CHANNEL))
}


def make_data_fit_session(plus, minus):
    session = CPFitSession(
        plus, minus, plus_data, minus_data, extended=True,
        signal_yield=Parameter("n_signal", N_SIGNAL, fixed=True),
    ).with_efficiency(plus_efficiency, minus_efficiency).with_veto(D0_VETO)
    for name, plus_pdf, minus_pdf, asymmetry, count in (
        ("qqbar", _qqbar_plus_pdf, _qqbar_minus_pdf, QQBAR_YIELD_ASYMMETRY, N_QQBAR),
        ("kpipibkg", _kpipibkg_pdf, _kpipibkg_pdf, KPIPIBKG_YIELD_ASYMMETRY, N_KPIPIBKG),
    ):
        session = session.with_background(
            name,
            plus_pdf.with_charge_asymmetry(plus.normalization_sample, veto=D0_VETO,
                                           asymmetry=asymmetry, charge=+1),
            minus_shape=minus_pdf.with_charge_asymmetry(
                minus.normalization_sample, veto=D0_VETO, asymmetry=asymmetry, charge=-1),
            yield_=Parameter(f"n_{name}", count, fixed=True),
        )
    return session


isobar_session = make_data_fit_session(isobar_models[+1], isobar_models[-1])
isobar_multifit = isobar_session.fit_multistart(
    n_starts=N_STARTS,
    seed=20260923,
    include_default=True,
    strategy=1,
    verbose=1,
    hessian="jax",
)
isobar_result = isobar_multifit.best
print(isobar_result.fmin)
if not isobar_result.valid:
    raise RuntimeError("Fit isobárico inválido: verificar MIGRAD/EDM antes de inicializar o QMI")
isobar_values = isobar_session.result_values(isobar_result)
isobar_session.print_result(isobar_result)
```

### Magnitude e fase ajustadas (fit isobárico)

Coeficientes `isobar_shared` no ponto de mínimo `isobar_values`, por carga: magnitude `|coef|` e fase relativa ao `rho0_770` da mesma carga (`arg(coef/rho)`), sem quadrado, raiz ou normalização individual -- mesma convenção da tabela de coeficientes globais usada mais abaixo para o modelo QMI ("Comparação com a tabela fornecida").

```python
def _wrap_deg(angle_deg):
    return (np.asarray(angle_deg) + 180.0) % 360.0 - 180.0


isobar_coefficient_rows = []
for name in isobar_shared:
    for charge, suffix in ((+1, "plus"), (-1, "minus")):
        coef = complex(isobar_shared[name].for_charge(charge).value(isobar_values))
        rho = complex(isobar_shared["rho0_770"].for_charge(charge).value(isobar_values))
        isobar_coefficient_rows.append({
            "component": name, "charge": suffix,
            "magnitude": abs(coef),
            "fase [deg]": float(_wrap_deg(np.degrees(np.angle(coef / rho)))),
        })
isobar_coefficient_table = pd.DataFrame(isobar_coefficient_rows).set_index(["component", "charge"])
display(isobar_coefficient_table)
```

## Inicialização QMI a partir do resultado isobárico


```python
def isobar_swave_single_pair(charge, mass):
    """Soma de um par, sem escalas individuais, com parâmetros do fit aos dados."""
    total = jnp.zeros_like(jnp.asarray(mass), dtype=jnp.complex128)
    for component in isobar_models[charge].amplitude_model.components:
        if component.name not in SWAVE_NAMES:
            continue
        dynamics = component.function
        total = total + component.coefficient.value(isobar_values) * dynamics.lineshape(
            jnp.asarray(mass), dynamics.context.resolve(isobar_values)
        )
    return np.asarray(total)


# QMI S-wave knots in m(pi+pi-), pair (2, 0) -- fixed list given for this fit.
QMI_KNOTS_GEV = (
    0.279140, 0.509015, 0.634185, 0.704998, 0.769832, 0.841871, 0.901300, 0.970000,
    0.980000, 0.990000, 1.000000, 1.070000, 1.133000, 1.195000, 1.250000, 1.310000,
    1.360000, 1.410000, 1.556780, 1.740000, 1.894000, 2.236000, 2.500000, 4.472130,
    4.817487, 5.139690,
)
# Linear interpolation in s=m(pi+pi-)**2, as requested for this comparison.
# Saved fit outputs must be regenerated after changing the interpolation.
QMI_INTERPOLATION = "none"  # Constant complex amplitude per mass bin.

# Optional QMI curvature penalty, as in 23_..._SqDP_QMI1D_Smoothness (docs/lineshapes.md,
# "Optional QMI knot smoothing"). QMISmoothnessConstraint rejects interpolation="none", so
# the same formula is applied to the constant bin values placed at the bin centers (see
# the session cell). NOT validated against toys here: lambda depends on the node scale
# (not comparable to notebook 23's) and post-fit errors come from the penalized objective.
# strength=0.0 is an exact no-op.
QMI_SMOOTHNESS_STRENGTH = 0.1  # lambda fixo; zero desliga a penalização
QMI_SMOOTHNESS_WEIGHTS = None  # None: only stencils entirely above the D0 veto are active;
                               # otherwise len(QMI_BIN_CENTERS)-2 weights, one per interior bin

# In step mode knots are EDGES; parameters belong to the 25 intervals.
QMI_BIN_CENTERS = (np.asarray(QMI_KNOTS_GEV[:-1]) + np.asarray(QMI_KNOTS_GEV[1:])) / 2
QMI_VETO_BINS = tuple(i for i, (lo, hi) in enumerate(zip(QMI_KNOTS_GEV[:-1], QMI_KNOTS_GEV[1:]))
                     if lo >= D0_VETO_MIN_GEV and hi <= D0_VETO_MAX_GEV)
QMI_FIXED_BINS = (0, *QMI_VETO_BINS)

_qmi_seed_amplitude = {
    charge: isobar_swave_single_pair(charge, QMI_BIN_CENTERS) for charge in (+1, -1)
}


_qmi_global_seed = {charge: complex(seed[0]) for charge, seed in _qmi_seed_amplitude.items()}
if any(not np.isfinite(value) or abs(value) < 1e-12 for value in _qmi_global_seed.values()):
    raise ValueError("Semente no primeiro nó é nula/não finita; não é possível fixá-lo em 1+0j")


from dataclasses import dataclass


@dataclass(frozen=True)
class CPBinPart:
    """Scalar x+charge*dx (or y+charge*dy), resolved by the dynamics API.

    Embedded Parameter.dynamics fields are discovered by DecayModel and
    shared across charges; resolution remains differentiable in JAX.
    """
    average: Parameter
    difference: Parameter
    charge: int

    def resolve(self, values=None):
        return self.average.resolve(values) + self.charge * self.difference.resolve(values)


normalized_seeds = {
    charge: seed / _qmi_global_seed[charge]
    for charge, seed in _qmi_seed_amplitude.items()
}
for seed in normalized_seeds.values():
    seed[0] = 1.0 + 0.0j
    seed[list(QMI_VETO_BINS)] = 0.0 + 0.0j
bin_average = (normalized_seeds[+1] + normalized_seeds[-1]) / 2
bin_difference = (normalized_seeds[+1] - normalized_seeds[-1]) / 2
QMI_BIN_CP = tuple(
    {
        key: Parameter.dynamics(
            f"S_QMI.bin_{i:02d}.{key}", float(value), owner="S_QMI",
            fixed=(i in QMI_FIXED_BINS),
            step=max(1e-5, 0.02 * max(abs(normalized_seeds[+1][i]), abs(normalized_seeds[-1][i]))),
        )
        for key, value in zip(("x", "y", "dx", "dy"),
                              (avg.real, avg.imag, diff.real, diff.imag))
    }
    for i, (avg, diff) in enumerate(zip(bin_average, bin_difference))
)


def make_qmi_lineshape(charge):
    return QMI(
        knots=QMI_KNOTS_GEV,
        real_parts=tuple(CPBinPart(p["x"], p["dx"], charge) for p in QMI_BIN_CP),
        imaginary_parts=tuple(CPBinPart(p["y"], p["dy"], charge) for p in QMI_BIN_CP),
        interpolation=QMI_INTERPOLATION,
    )


S_QMI_LINESHAPES = {charge: make_qmi_lineshape(charge) for charge in (+1, -1)}
shared_coefficients = make_shared_coefficients(isobar_values)
# c_plus = (x+dx) + i(y+dy); c_minus = (x-dx) + i(y-dy).
_c_average = (_qmi_global_seed[+1] + _qmi_global_seed[-1]) / 2
_c_difference = (_qmi_global_seed[+1] - _qmi_global_seed[-1]) / 2
shared_coefficients["S_QMI"] = CPRealImag(*(
    Parameter.coefficient(f"S_QMI.{key}", float(value), owner="S_QMI", fixed=False)
    for key, value in zip(("x", "y", "dx", "dy"),
                          (_c_average.real, _c_average.imag, _c_difference.real, _c_difference.imag))
))
dynamics_parameters = make_dynamics_parameters(isobar_values)
print(f"Independent QMI: {len(QMI_KNOTS_GEV)} masses, {4 * (len(QMI_BIN_CENTERS) - len(QMI_FIXED_BINS))} free node parameters + 4 global CP parameters")
```

```python
start_mass_scan = np.linspace(QMI_KNOTS_GEV[0], QMI_KNOTS_GEV[-1], 4000)
start_swave_curves = {}
fig_start_swave, start_axes = plt.subplots(
    2, 2, figsize=(12, 7), sharex=True, sharey="row", constrained_layout=True,
)

for col, (charge, charge_label) in enumerate(((+1, "B+"), (-1, "B−"))):
    qmi_parameters = S_QMI_LINESHAPES[charge]
    start_real = np.asarray([p.resolve({}) for p in qmi_parameters.real_parts])
    start_imag = np.asarray([p.resolve({}) for p in qmi_parameters.imaginary_parts])
    start_qmi = QMI(
        knots=qmi_parameters.knots,
        real_parts=tuple(start_real), imaginary_parts=tuple(start_imag),
        interpolation=qmi_parameters.interpolation,
    )
    start_magnitude, start_phase = start_qmi.interpolated_magnitude_phase(jnp.asarray(start_mass_scan))
    global_coefficient = complex(shared_coefficients["S_QMI"].for_charge(charge).value({}))
    reference_phase = float(jnp.angle(shared_coefficients["rho0_770"].for_charge(charge).value({})))
    # Cartesian nodes make interpolated_magnitude_phase recompute jnp.angle(re + i*im) at
    # every scan point, wrapped to (-pi, pi]; unwrap so the plotted curve stays continuous
    # instead of showing spurious ~2*pi jumps wherever the physical phase crosses the cut.
    start_phase = np.unwrap(np.angle(global_coefficient * np.exp(1j * np.asarray(start_phase)))) - reference_phase
    start_intensity = abs(global_coefficient) ** 2 * np.asarray(start_magnitude) ** 2
    start_swave_curves[charge_label] = {
        "mass_GeV": start_mass_scan, "intensity": start_intensity, "phase_rad": start_phase,
    }
    start_node_intensity = abs(global_coefficient) ** 2 * (start_real ** 2 + start_imag ** 2)
    # Unwrap across knots (increasing mass order) so markers land on the unwrapped curve
    # above instead of off it by a multiple of 2*pi at whichever knot the wrap cut falls on.
    start_node_phase = np.unwrap(np.angle(global_coefficient * (start_real + 1j * start_imag))) - reference_phase

    start_axes[0, col].plot(start_mass_scan, start_intensity, label="QMI de start")
    start_axes[0, col].scatter(QMI_BIN_CENTERS, start_node_intensity,
                               s=18, color="black", zorder=3, label="Nós iniciais")
    start_axes[1, col].plot(start_mass_scan, start_phase)
    start_axes[1, col].scatter(QMI_BIN_CENTERS, start_node_phase,
                               s=18, color="black", zorder=3)
    for row in (0, 1):
        start_axes[row, col].axvspan(D0_VETO_MIN_GEV, D0_VETO_MAX_GEV,
                                    color="grey", alpha=0.15, label="Veto D0")
    start_axes[0, col].set_title(f"{charge_label}: start QMI ({qmi_parameters.interpolation})")
    start_axes[0, col].legend()
    start_axes[1, col].set_xlabel(r"$m(\pi^+\pi^-)$ [GeV]")

start_axes[0, 0].set_ylabel(r"$|A_S^{\mathrm{start}}|^2$ (unidades arbitrárias)")
start_axes[1, 0].set_ylabel(r"Fase relativa ao $\rho$ [rad]")
plt.show()
```

## Componentes do modelo QMI

Os seis isóbaros restantes e `S_QMI` usam `normalize_component=False`; ambos os
modelos usam `normalize_components=False`, assim como no fit isobárico anterior.
O coeficiente externo da QMI é CPRealImag livre; o primeiro bin é fixo em `1+0j` por carga.
A densidade total continua normalizada conjuntamente para B+ e B−, incluindo interferências.


```python
def build_components(charge, shared, dynamics):
    coefficient = {name: cp.for_charge(charge) for name, cp in shared.items()}
    components = [
        Resonance(
            name, RESONANCE_PAIR, coefficient[name],
            lineshape=LINESHAPES[name](), angular=ZemachP(),
            normalize_component=False,
            mass=dynamics[f"{name}.mass"], width=dynamics[f"{name}.width"], spin=SPINS[name],
            resonance_radius=4.0, parent_radius=4.0,
        )
        for name in LINESHAPES if name not in SWAVE_NAMES
    ]
    components.append(
        Resonance(
            "S_QMI", RESONANCE_PAIR, coefficient["S_QMI"],
            lineshape=S_QMI_LINESHAPES[charge], angular=ZemachP(),
            normalize_component=False,
            mass=1.0, width=0.0, spin=0,
            resonance_radius=4.0, parent_radius=4.0,
        )
    )
    return components


plus_model = DecayModel(
    PLUS_CHANNEL, build_components(+1, shared_coefficients, dynamics_parameters),
    normalize_components=False,
    **NORMALIZATION_CONFIG
)
minus_model = DecayModel(
    MINUS_CHANNEL, build_components(-1, shared_coefficients, dynamics_parameters),
    normalize_components=False,
    **NORMALIZATION_CONFIG
)

print("B+ components:", [c.name for c in plus_model.amplitude_model.components])
print("B+ normalization sample size:", plus_model.normalization_sample.size)
```

com `yield_=0` fixo (inerte, ver acima).

```python
fit_session = make_data_fit_session(plus_model, minus_model)

# Step QMI: bin values sit at the bin centers of an auxiliary "linear" QMI that shares the
# same CPBinPart parameters. It only feeds the penalty; the fitted amplitude stays constant
# per bin. By default only the region after the D0 veto is smoothed: weight k is the
# stencil of bins (k, k+1, k+2), active only if all three lie above the veto bin (fixed at 0).
if QMI_SMOOTHNESS_WEIGHTS is None:
    QMI_SMOOTHNESS_WEIGHTS = tuple(
        1.0 if k > max(QMI_VETO_BINS) else 0.0 for k in range(len(QMI_BIN_CENTERS) - 2)
    )


def qmi_bin_smoothness(charge):
    step_qmi = S_QMI_LINESHAPES[charge]
    centers_qmi = QMI(knots=tuple(float(m) for m in QMI_BIN_CENTERS),
                      real_parts=step_qmi.real_parts, imaginary_parts=step_qmi.imaginary_parts,
                      interpolation="linear")
    return centers_qmi.smoothness_constraint(strength=QMI_SMOOTHNESS_STRENGTH,
                                             weights=QMI_SMOOTHNESS_WEIGHTS)


# Preserve the original penalty on the physical nodes c_q * QMI_q.
# The curvature penalty is quadratic, so multiply the shape penalty by |c_q|².
def physical_qmi_smoothness(charge):
    shape_penalty = qmi_bin_smoothness(charge)
    coefficient = shared_coefficients["S_QMI"].for_charge(charge)

    def penalty(fit_values):
        return jnp.abs(coefficient.value(fit_values)) ** 2 * shape_penalty(fit_values)

    return penalty


QMI_SMOOTHNESS_CONSTRAINTS = tuple(physical_qmi_smoothness(charge) for charge in (+1, -1))
for penalty in QMI_SMOOTHNESS_CONSTRAINTS:
    fit_session = fit_session.with_constraint(penalty)

n_free = sum(1 for p in fit_session.parameters if not p.fixed)
n_fixed = sum(1 for p in fit_session.parameters if p.fixed)
print(f"parameters: {len(fit_session.parameters)} total, {n_free} free, {n_fixed} fixed")
print("S_QMI: 25 constant bins (re+im) per charge; first bin fixed at 1+0j and D0 bin fixed at 0; global x/y/dx/dy free")
print(f"QMI smoothness strength: {QMI_SMOOTHNESS_STRENGTH} (unvalidated -- see cell above)")
print(f"QMI smoothness weights: {QMI_SMOOTHNESS_WEIGHTS}")
assert n_free == 4 * (len(QMI_BIN_CENTERS) - len(QMI_FIXED_BINS)) + 4 + 21  # shape nodes + global S CP + other coefficients
```

## Fit fractions físicas por carga

Integrais no DP completo, sem eficiência/veto, com `integral(f)=mean(weights*f)`.
A FF da onda-S é a da amplitude QMI simetrizada, sem chi_c0. Comparar com uma soma
isobárica exige incluir suas interferências. A FF CP média é ponderada pelas taxas R±.

```python
component_names = [c.name for c in plus_model.amplitude_model.components]
# Physical caches exclude efficiency/veto; accepted charge fractions must not replace these rates.
physical_caches = {
    charge: model.prepare_cache(model.normalization_sample.take(jnp.arange(1)))
    for charge, model in (("plus", plus_model), ("minus", minus_model))
}


def physical_component_rates(cache, fit_values):
    matrix = cache.normalization_matrix(fit_values)
    coefficients = cache.coefficient_vector(fit_values)
    rates = jnp.abs(coefficients) ** 2 * jnp.real(jnp.diag(matrix))
    total = jnp.real(jnp.vdot(coefficients, matrix @ coefficients))
    return rates, total


physical_rates = {q: physical_component_rates(cache, values) for q, cache in physical_caches.items()}
fit_fractions = {
    q: dict(zip(component_names, map(float, rates / total)))
    for q, (rates, total) in physical_rates.items()
}
physical_ff_cp = dict(zip(component_names, map(float,
    (physical_rates["plus"][0] + physical_rates["minus"][0])
    / (physical_rates["plus"][1] + physical_rates["minus"][1]))))
print("R_plus =", float(physical_rates["plus"][1]), "R_minus =", float(physical_rates["minus"][1]))
display(pd.DataFrame(fit_fractions) * 100.0)
```

```python
def _ff_asymmetry_percent(plus_ff, minus_ff):
    return {
        name: 100.0 * (minus_ff[name] - plus_ff[name]) / (minus_ff[name] + plus_ff[name])
        for name in plus_ff
    }


component_names = [c.name for c in plus_model.amplitude_model.components]
plus_ff_pct = {n: 100.0 * fit_fractions["plus"][n] for n in component_names}
minus_ff_pct = {n: 100.0 * fit_fractions["minus"][n] for n in component_names}

our_table = pd.DataFrame({
    "B- FF [%]": minus_ff_pct,
    "B+ FF [%]": plus_ff_pct,
    "FF asym [%]": _ff_asymmetry_percent(plus_ff_pct, minus_ff_pct),
})
our_table
```

### ACP integrado por componente

`ACP = (I_minus - I_plus)/(I_minus + I_plus)`, com as integrais físicas de cada
componente. Para a QMI independente, entram os nós de ambas as cargas. A incerteza
propaga a covariância conjunta de todos os parâmetros livres, incluindo dinâmica.

```python
def component_acp_percent(fit_values):
    plus_rates, _ = physical_component_rates(physical_caches["plus"], fit_values)
    minus_rates, _ = physical_component_rates(physical_caches["minus"], fit_values)
    return 100.0 * (minus_rates - plus_rates) / (minus_rates + plus_rates)


free_parameter_names = [p.name for p in fit_session.parameters if not p.fixed]
acp_errors = delta_method_errors(component_acp_percent, values, free_parameter_names, result.covariance)
acp_table = pd.DataFrame({
    "ACP [%]": np.asarray(component_acp_percent(values)),
    "sigma(ACP)": np.asarray(acp_errors),
}, index=component_names)
display(acp_table)
# These extra caches are only needed for physical observable derivatives, not fit evaluation.
del physical_caches
```

## Comparação com a tabela fornecida: coeficientes globais (NLL 967471)

Os valores de `A_i` são tratados como **magnitudes dos coeficientes globais**,

## Onda-S por carga: magnitude e fase, com e sem chi_c0

Curvas de um único par em função de m(pi+pi-), sem construir pontos artificiais do DP.
A QMI é avaliada diretamente nos seus nós ajustados. chi_c0 entra com sua lineshape
bruta e coeficiente, sem normalização individual, como no ajuste. A fase é relativa ao rho de cada carga.
Os eixos de magnitude têm a mesma escala em B+ e B−; a intensidade de uma única
lineshape não é uma projeção de contagens nem a integral física da onda-S.

```python
def swave_scan(model, values, mass_scan, *, include_chic0):
    charge = +1 if model is plus_model else -1
    label = "plus" if charge == +1 else "minus"
    real_parts = tuple(p.resolve(values) for p in S_QMI_LINESHAPES[charge].real_parts)
    imaginary_parts = tuple(p.resolve(values) for p in S_QMI_LINESHAPES[charge].imaginary_parts)
    qmi = QMI(knots=QMI_KNOTS_GEV, real_parts=real_parts, imaginary_parts=imaginary_parts,
              interpolation=QMI_INTERPOLATION)
    mag, phase = qmi.interpolated_magnitude_phase(jnp.asarray(mass_scan))
    amplitude = coefficient_at("S_QMI", charge, values) * mag * jnp.exp(1j * phase)
    if include_chic0:
        component = next(c for c in model.amplitude_model.components if c.name == "chi_c0")
        dynamics = component.function
        amplitude = amplitude + (coefficient_at("chi_c0", charge, values)
            * dynamics.lineshape(jnp.asarray(mass_scan), dynamics.context.resolve(values)))
    reference = coefficient_at("rho0_770", charge, values)
    return np.asarray(amplitude * jnp.exp(-1j * jnp.angle(reference)))


swave_mass_scan = np.linspace(QMI_KNOTS_GEV[0], QMI_KNOTS_GEV[-1], 4000)

swave_curves = {}
for charge_label, charge_model in (("B+", plus_model), ("B-", minus_model)):
    for include_chic0, tag in ((False, "S_QMI only"), (True, "S_QMI + chi_c0")):
        swave_curves[(charge_label, tag)] = swave_scan(
            charge_model, values, swave_mass_scan, include_chic0=include_chic0,
        )

qmi_only_knot_values = {
    charge_label: swave_scan(charge_model, values, QMI_BIN_CENTERS, include_chic0=False)
    for charge_label, charge_model in (("B+", plus_model), ("B-", minus_model))
}

fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharex=True, sharey="row")
for col, charge_label in enumerate(("B+", "B-")):
    for tag, style in (("S_QMI only", "--"), ("S_QMI + chi_c0", "-")):
        amplitude = swave_curves[(charge_label, tag)]
        axes[0, col].plot(swave_mass_scan, np.abs(amplitude) ** 2, style, label=tag)
        axes[1, col].plot(swave_mass_scan, np.unwrap(np.angle(amplitude)), style, label=tag)

    axes[0, col].scatter(
        QMI_BIN_CENTERS, np.abs(qmi_only_knot_values[charge_label]) ** 2,
        s=10, color="black", zorder=3,
    )
    # Unwrap the knot phases and shift each onto the same 2*pi branch as the unwrapped
    # "S_QMI only" curve, so the markers land on the plotted line.
    qmi_only_phase_curve = np.unwrap(np.angle(swave_curves[(charge_label, "S_QMI only")]))
    knot_phase = np.unwrap(np.angle(qmi_only_knot_values[charge_label]))
    curve_at_knots = np.interp(QMI_BIN_CENTERS, swave_mass_scan, qmi_only_phase_curve)
    knot_phase = knot_phase + 2 * np.pi * np.round((curve_at_knots - knot_phase) / (2 * np.pi))
    axes[1, col].scatter(QMI_BIN_CENTERS, knot_phase, s=10, color="black", zorder=3)
    for mass in (D0_VETO_MIN_GEV, D0_VETO_MAX_GEV):
        axes[0, col].axvline(mass, color="grey", ls=":", lw=1)
        axes[1, col].axvline(mass, color="grey", ls=":", lw=1)

    axes[0, col].set_title(f"{charge_label} onda-S")
    axes[1, col].set_xlabel(r"$m(\pi^+\pi^-)$ [GeV]")
    axes[0, col].legend()

axes[0, 0].set_ylabel(r"$|A_S|^2$ (unidades arbitrárias)")
axes[1, 0].set_ylabel(r"fase$(A_S)$ [rad]")
fig.tight_layout()
plt.show()
```

### Onda-S antes do veto D0

Mesma comparação acima (QMI sem `chi_c0` vs. Tabela 203-112), restrita aos bins com

```python
def plot_swave_region(region, region_title):
    """Comparação QMI vs. Tabela 203-112 restrita a um bloco contíguo de bins."""
    bins = np.flatnonzero(region)
    assert np.all(np.diff(bins) == 1), "a região deve ser um bloco contíguo de bins"
    region_edges = TABLE_MASS_EDGES_GEV[bins[0]: bins[-1] + 2]
    region_centers = centers[region]

    fig, axes = plt.subplots(2, 2, figsize=(13, 8), sharex=True, sharey="row")
    for col, (label, model) in enumerate(zip(("B+", "B-"), models_for_comparison)):
        midpoint_amp = swave_scan(model, fitted_values_for_comparison, region_centers, include_chic0=False)
        sampled_amp = np.asarray([
            swave_scan(model, sample, region_centers, include_chic0=False)
            for sample in parameter_samples
        ])

        fit_intensity = np.abs(midpoint_amp) ** 2 / fit_scale
        intensity_low, intensity_high = np.nanpercentile(np.abs(sampled_amp) ** 2 / fit_scale, (16, 84), axis=0)
        axes[0, col].errorbar(
            region_centers, fit_intensity,
            yerr=np.vstack((fit_intensity - intensity_low, intensity_high - fit_intensity)), fmt="none",
            ecolor="tab:blue", elinewidth=1.2, capsize=2, label="incerteza 68% do fit",
        )
        axes[0, col].stairs(fit_intensity, region_edges, baseline=None, color="tab:blue",
                            linewidth=1.6, label="QMI ajustado (centro de cada bin)")
        axes[0, col].stairs(reference_intensity[region, col] / table_scale, region_edges,
                            baseline=None, color="tab:red", linewidth=1.6,
                            label="Tabela 203-112 (constante por bin)")

        fit_phase = np.degrees(np.unwrap(np.angle(midpoint_amp)))
        phase_delta = np.degrees(np.unwrap(np.angle(sampled_amp), axis=1)) - fit_phase[None, :]
        phase_low, phase_high = np.nanpercentile(phase_delta, (16, 84), axis=0)
        axes[1, col].errorbar(
            region_centers, fit_phase, yerr=np.vstack((-phase_low, phase_high)), fmt="none",
            ecolor="tab:blue", elinewidth=1.2, capsize=2, label="incerteza 68% do fit",
        )
        axes[1, col].stairs(fit_phase, region_edges, baseline=None, color="tab:blue",
                            linewidth=1.6, label="QMI ajustado (fase desenrolada)")
        table_phase = np.degrees(np.unwrap(np.radians(
            TABLE_SWAVE[region, 2 * col + 1] + TABLE_PHASE_OFFSET_DEG[label]
        )))
        valid_phase = reference_intensity[region, col] > 0
        axes[1, col].stairs(np.where(valid_phase, table_phase, np.nan), region_edges,
                            baseline=None, color="tab:red", linewidth=1.6,
                            label="Tabela 203-112 (fase desenrolada)")

        axes[0, col].set_title(f"{label}: {region_title} ({COMPARISON_MODE})")
        axes[1, col].set_xlabel(r"$m(\pi^+\pi^-)$ [GeV]")
        axes[1, col].set_xlim(region_edges[0], region_edges[-1])
        for ax in axes[:, col]:
            ax.legend(fontsize=8)
    axes[0, 0].set_ylabel(r"$|A|^2 / \int (|A_+|^2+|A_-|^2)\,dm$ [GeV$^{-1}$]"
                          if COMPARISON_MODE == "shape" else r"$|A|^2$ (escala bruta)")
    axes[1, 0].set_ylabel("Fase [graus, desenrolada]")
    fig.tight_layout()
    plt.show()


plot_swave_region(right <= D0_VETO_MIN_GEV, "onda-S antes do veto D0")
```

```python
from scipy.stats import norm

# 1D pull distribution of the non-empty DP bins, with an unbinned ML Gaussian fit
fig, axes = plt.subplots(1, 2, figsize=(12, 4), constrained_layout=True)
bins_1d = np.linspace(-pull_max, pull_max, 31)
x_grid = np.linspace(-pull_max, pull_max, 400)
for ax, (charge, res) in zip(axes, gof_dp.items()):
    pulls = np.asarray(res.pulls).ravel()
    pulls = pulls[np.isfinite(pulls)]
    n = pulls.size
    mu, sigma = norm.fit(pulls)
    mu_err, sigma_err = sigma / np.sqrt(n), sigma / np.sqrt(2 * n)
    bin_width = bins_1d[1] - bins_1d[0]
    counts, _ = np.histogram(pulls, bins=bins_1d)
    centers = 0.5 * (bins_1d[:-1] + bins_1d[1:])
    ax.errorbar(centers, counts, yerr=np.sqrt(counts), fmt="o", color="k", ms=4, label=f"bins ({n})")
    ax.plot(x_grid, n * bin_width * norm.pdf(x_grid, mu, sigma), "r-",
            label=rf"fit: $\mu={mu:.2f}\pm{mu_err:.2f}$, $\sigma={sigma:.2f}\pm{sigma_err:.2f}$")
    ax.plot(x_grid, n * bin_width * norm.pdf(x_grid, 0.0, 1.0), "b--", label=r"$\mathcal{N}(0,1)$")
    ax.set_xlabel(r"pull $(o-e)/\sqrt{e}$")
    ax.set_ylabel("bins")
    ax.set_title(f"B{'+' if charge == 'plus' else '-'} DP pull distribution")
    ax.legend(fontsize=9)
```

```python
# Resíduos locais adaptativos por k-vizinhos (KD-tree)
from dalitzplotfitter.kinematics import invariants_to_square_dalitz, fold_thetaprime

def _kdtree_points(charge, square=False, k=50, size=120_000):
    data = fit_session.plus_data if charge == "plus" else fit_session.minus_data
    sample = fit_session.plus_model.generate_phase_space(size, seed=20260911 if charge == "plus" else 20260912)
    other = fit_session.minus_model.generate_phase_space(size, seed=20260913 if charge == "plus" else 20260914)
    comps = fit_session._projection_components_pair(fit_session.result_values(result), sample, other)[0 if charge == "plus" else 1]
    exp_points, exp_weights = [], []
    for _, s, w in comps:
        d = s.as_dict()
        if square:
            a, b = invariants_to_square_dalitz(d["s12"], d["s13"], d["s23"], mother_mass=MOTHER_MASS, masses=DAUGHTER_MASSES, pair=SQDP_PAIR)
            b = fold_thetaprime(b)
        else:
            a, b = jnp.minimum(d["s13"], d["s23"]), jnp.maximum(d["s13"], d["s23"])
        exp_points.append(np.column_stack((np.asarray(a), np.asarray(b))))
        exp_weights.append(np.asarray(w))
    d = data.as_dict()
    if square:
        a, b = invariants_to_square_dalitz(d["s12"], d["s13"], d["s23"], mother_mass=MOTHER_MASS, masses=DAUGHTER_MASSES, pair=SQDP_PAIR)
        b = fold_thetaprime(b)
    else:
        a, b = jnp.minimum(d["s13"], d["s23"]), jnp.maximum(d["s13"], d["s23"])
    return kdtree_local_residuals(np.column_stack((np.asarray(a), np.asarray(b))), np.concatenate(exp_points), np.concatenate(exp_weights), k=k)

kdtree_residuals = {}
for coordinate in ("dp", "square"):
    for charge in ("plus", "minus"):
        pulls, observed_count, expected_count, radii = _kdtree_points(charge, square=coordinate == "square")
        kdtree_residuals[(coordinate, charge)] = pulls
        print(coordinate, charge, "median pull=", float(np.median(pulls)), "RMS=", float(np.std(pulls)))
```

# Parte 3 — Ajustes QMI2D e comparação com o isobárico e o QMI 1D

A onda-S (`S_QMI`, ou `sigma0 + f0_980 + f0_1370` no isobárico) é substituída por um campo complexo
**bidimensional**, `DalitzAmplitude("S_QMI2D", QMI2D(...))`, constante por célula
(`interpolation="none"`, análogo à QMI 1D em degraus). Os outros seis componentes
(`rho0_770`, `omega_782`, `rho0_1450`, `f2_1270`, `rho0_3_1690`, `chi_c0`), os dados, a eficiência,
o veto D0, os backgrounds, os yields e a quadratura são os mesmos dos dois ajustes anteriores.

- **Coordenadas:** a ordem das filhas é `(pi±, pi±, pi∓)`, então os dois pares `pi+pi-` são `s13` e `s23`.
  O adaptador `OppositeSignQMI2D` passa `(s13, s23)` ao `QMI2D(folded=True)`, que usa
  `s_low = min(s13, s23)` e `s_high = max(s13, s23)`. Não há fator extra de simetrização.
- **Células:** bordas comuns nos dois eixos, finas até o veto D0 (entre 1.0 e 1.74 GeV, nós do QMI 1D),
  com uma borda em cada extremo do veto.
  Células fora do Dalitz físico (`physical_bin_mask`) ou inteiramente no veto são inativas (amplitude 0).
- **CP por célula**, como os bins da QMI 1D: `z_q = (x + q dx) + i (y + q dy)` por célula, convertido para
  magnitude/fase pelo helper `CPCellPolar`. `QMI2D_PER_CELL_CP = False` fixa `dx = dy = 0` (só CPV global).
- **Células pouco povoadas** (`min(n+, n-) < QMI2D_MIN_EVENTS`) compartilham os parâmetros da célula
  bem povoada mais próxima (preferindo a mesma faixa de `s_low`), o que equivale a fundi-las;
  sem isso, células de fronteira com 1–15 candidatos deixariam direções mal determinadas na Hessiana.
- **Convenção:** a célula âncora (a mais povoada) é fixa em `1+0j` nas duas cargas; o coeficiente global
  `S_QMI2D.x/y/dx/dy` é livre, como `S_QMI` na parte 2. `rho0_770` continua sendo a referência global.
- **Semente:** a onda-S do ajuste QMI 1D (coeficiente × QMI simetrizada), por carga, no ponto médio dos
  candidatos de cada célula (um ponto físico, fora do veto). Os demais coeficientes partem do resultado QMI 1D.

**Dois ajustes 2D, na mesma grade:** `QMI 2D` (todos os coeficientes livres) e `QMI 2D ρ(1450) fixo`
(`rho0_1450.x/y/dx/dy` fixos no resultado do QMI 1D). O segundo mede quanto o campo compete com o
ρ(1450): o ΔNLL entre os dois é o custo de impor o ρ(1450) do QMI 1D.

**Interpretação:** um campo 2D livre não é uma onda-S pura. Ele pode absorver estrutura angular
e competir com as ressonâncias largas explícitas (em especial `rho0_1450`). As frações e a ACP do
`S_QMI2D` referem-se ao campo inteiro; a comparação com a linha "S-wave" das referências é indicativa.
Magnitude/fase do coeficiente global `S_QMI2D` dependem da escolha da âncora e não são comparáveis
com as da QMI 1D nem com as das tabelas.


```python
from dataclasses import dataclass
from types import SimpleNamespace

from dalitzplotfitter import DalitzAmplitude, QMI2D, dalitz_s13_limits, physical_bin_mask

# Bordas comuns em s (GeV²) para s_low = min(s13, s23) e s_high = max(s13, s23).
# Finas até o veto D0: sigma e f0(980) abaixo de 1 GeV e, entre 1.0 e 1.74 GeV (região do rho(1450)),
# nós exatos de QMI_KNOTS_GEV (um sim, um não). Uma grade com só 1.2 e 1.5 GeV nessa faixa deixava a
# onda-S mal descrita e o ajuste zerava o rho(1450) em B+ (NLL 114.5 pior). Uma borda em cada extremo
# do veto e células largas na região central, pouco povoada. Extremos = limites cinemáticos exatos.
QMI2D_KNOTS_FROM_1D_GEV = tuple(k for k in QMI_KNOTS_GEV if round(k, 3) in (1.07, 1.195, 1.31, 1.41, 1.557))
assert len(QMI2D_KNOTS_FROM_1D_GEV) == 5
QMI2D_EDGES_GEV2 = (
    (DAUGHTER_MASSES[0] + DAUGHTER_MASSES[1]) ** 2,
    0.30, 0.55, 0.81, 0.941, 1.00,
    *(m ** 2 for m in QMI2D_KNOTS_FROM_1D_GEV),
    D0_VETO_MIN_GEV ** 2, D0_VETO_MAX_GEV ** 2,
    7.0, 14.0,
    (MOTHER_MASS - DAUGHTER_MASSES[2]) ** 2,
)
print("bordas (massa, GeV):", np.round(np.sqrt(QMI2D_EDGES_GEV2), 3))
QMI2D_INTERPOLATION = "none"  # constante por célula, como a QMI 1D em degraus acima
QMI2D_PER_CELL_CP = True      # x, y, dx, dy por célula; False: dx = dy = 0 (só CPV global)
QMI2D_MIN_EVENTS = 50         # min(n+, n-) abaixo disso: célula compartilha os parâmetros
                              # da célula bem povoada mais próxima (mesma faixa de s_low, se possível)
QMI2D_ANCHOR_CELL = None      # None: célula ativa mais povoada; fixa em 1+0j nas duas cargas


def _folded(sample):
    return (np.minimum(np.asarray(sample.s13), np.asarray(sample.s23)),
            np.maximum(np.asarray(sample.s13), np.asarray(sample.s23)))


def make_qmi2d_grid(edges, *, min_events=QMI2D_MIN_EVENTS, anchor=None):
    """Máscara física/veto, ocupação, âncora, fusão de células pouco povoadas e pontos representativos."""
    if min_events <= 0:
        raise ValueError("min_events deve ser positivo: toda célula dona precisa de candidatos")
    edges_array = np.asarray(edges, dtype=float)
    shape = (len(edges_array) - 1,) * 2
    mask_options = dict(mother_mass=MOTHER_MASS, masses=DAUGHTER_MASSES, folded=True)
    mask = np.asarray(physical_bin_mask(tuple(edges), tuple(edges), samples_per_bin=257, **mask_options))
    assert np.array_equal(mask, np.asarray(physical_bin_mask(
        tuple(edges), tuple(edges), samples_per_bin=1025, **mask_options))), "Refinar a amostragem da máscara física."
    mask &= np.triu(np.ones(shape, dtype=bool))  # plano dobrado: s_low <= s_high
    # Linha/coluna inteiramente dentro do veto D0: sem dados nem normalização -> inativa (amplitude 0).
    veto_bins = [k for k in range(shape[0])
                 if edges_array[k] >= D0_VETO_MIN_GEV ** 2 - 1e-12
                 and edges_array[k + 1] <= D0_VETO_MAX_GEV ** 2 + 1e-12]
    mask[veto_bins, :] = False
    mask[:, veto_bins] = False

    occupancy = {
        label: np.histogram2d(*_folded(sample), bins=(edges_array, edges_array))[0].astype(int)
        for label, sample in (("plus", plus_data), ("minus", minus_data))
    }
    for label, counts in occupancy.items():
        if counts[~mask].sum():
            raise ValueError(f"{label}: {counts[~mask].sum()} eventos em células inativas")
    min_counts = np.minimum(occupancy["plus"], occupancy["minus"])

    active_cells = [tuple(map(int, c)) for c in np.argwhere(mask)]
    if anchor is None:
        anchor = max(active_cells, key=lambda c: min_counts[c])
    free_cells = [c for c in active_cells if min_counts[c] >= min_events]
    assert anchor in free_cells

    def owner_of(cell):
        """Célula dona dos parâmetros: ela própria ou a vizinha bem povoada mais próxima."""
        if cell in free_cells:
            return cell
        i, j = cell
        return min(free_cells, key=lambda c: (4 * (c[0] - i) ** 2 + (c[1] - j) ** 2, c))

    # Ponto representativo de cada célula: média dos candidatos das duas cargas nela. O domínio
    # de Dalitz dobrado recortado por um retângulo é convexo, então o ponto é físico e fora do veto.
    # Células sem candidatos não têm ponto; elas nunca são donas (min_events > 0), só herdam parâmetros.
    all_low, all_high = (np.concatenate(pair) for pair in zip(_folded(plus_data), _folded(minus_data)))
    index = (np.searchsorted(edges_array, all_low, side="right") - 1,
             np.searchsorted(edges_array, all_high, side="right") - 1)
    points = {}
    for cell in active_cells:
        in_cell = (index[0] == cell[0]) & (index[1] == cell[1])
        if in_cell.any():
            points[cell] = (float(all_low[in_cell].mean()), float(all_high[in_cell].mean()))

    return SimpleNamespace(edges=tuple(edges), edges_array=edges_array, shape=shape, mask=mask,
                           occupancy=occupancy, min_counts=min_counts, active_cells=active_cells,
                           anchor=anchor, free_cells=free_cells,
                           owner={cell: owner_of(cell) for cell in active_cells}, points=points)


def describe_qmi2d_grid(grid):
    table = pd.DataFrame([
        dict(i=i, j=j, s_low_min=grid.edges_array[i], s_low_max=grid.edges_array[i + 1],
             s_high_min=grid.edges_array[j], s_high_max=grid.edges_array[j + 1],
             n_plus=grid.occupancy["plus"][i, j], n_minus=grid.occupancy["minus"][i, j],
             owner=grid.owner[(i, j)], anchor=(i, j) == grid.anchor)
        for i, j in grid.active_cells
    ])
    params_per_cell = 4 if QMI2D_PER_CELL_CP else 2
    print(f"Células ativas: {len(grid.active_cells)}   grupos de parâmetros: {len(grid.free_cells)}   "
          f"âncora: {grid.anchor}")
    print(f"Parâmetros livres do campo: {params_per_cell * (len(grid.free_cells) - 1)} "
          f"(+4 do coeficiente global S_QMI2D)")
    display(table[table.owner != table[["i", "j"]].apply(tuple, axis=1)])
    return table


# Fronteira exata do Dalitz dobrado, para distinguir a célula da sua parte física.
_q2d_bx = np.linspace((2 * DAUGHTER_MASSES[0]) ** 2 + 1e-9, (MOTHER_MASS - DAUGHTER_MASSES[2]) ** 2 - 1e-9, 3000)
_q2d_blo, _q2d_bhi = map(np.asarray, dalitz_s13_limits(
    jnp.asarray(_q2d_bx), mother_mass=MOTHER_MASS, masses=DAUGHTER_MASSES))
_q2d_blo = np.maximum(_q2d_blo, _q2d_bx)
_q2d_bvalid = _q2d_bhi >= _q2d_blo


def draw_q2d_boundary(ax):
    ax.plot(_q2d_bx[_q2d_bvalid], _q2d_blo[_q2d_bvalid], "k-", lw=0.8)
    ax.plot(_q2d_bx[_q2d_bvalid], _q2d_bhi[_q2d_bvalid], "k-", lw=0.8)
    for edge in (D0_VETO_MIN_GEV ** 2, D0_VETO_MAX_GEV ** 2):
        ax.axvline(edge, color="grey", ls=":", lw=0.8)
        ax.axhline(edge, color="grey", ls=":", lw=0.8)


def plot_qmi2d_grid(grid, title=""):
    fig, axes = plt.subplots(1, 2, figsize=(13, 5.5), constrained_layout=True)
    for ax, label in zip(axes, ("plus", "minus")):
        mesh = ax.pcolormesh(grid.edges_array, grid.edges_array,
                             np.ma.array(grid.occupancy[label], mask=~grid.mask).T,
                             shading="flat", edgecolors="grey", linewidth=0.3, norm="log")
        for cell, owner in grid.owner.items():
            if owner != cell and cell in grid.points:
                ax.annotate("", xy=grid.points[owner], xytext=grid.points[cell],
                            arrowprops=dict(arrowstyle="->", color="white", lw=1.2))
        ax.scatter(*grid.points[grid.anchor], marker="*", s=150, color="red", zorder=4, label="âncora (1+0j)")
        draw_q2d_boundary(ax)
        ax.set(xscale="log", yscale="log", xlabel=r"$s_\mathrm{low}$ [GeV$^2$]",
               ylabel=r"$s_\mathrm{high}$ [GeV$^2$]",
               title=f"{title}B{'+' if label == 'plus' else '-'}: candidatos por célula")
        ax.legend(loc="lower right")
        fig.colorbar(mesh, ax=ax, label="candidatos / célula")
    plt.show()


q2d_grid = make_qmi2d_grid(QMI2D_EDGES_GEV2, anchor=QMI2D_ANCHOR_CELL)
# Nomes usados pelas células seguintes (mapas, projeções).
_q2d_edges, _q2d_shape, _q2d_mask = q2d_grid.edges_array, q2d_grid.shape, q2d_grid.mask
qmi2d_occupancy, QMI2D_CELL_OWNER, QMI2D_CELL_POINTS = q2d_grid.occupancy, q2d_grid.owner, q2d_grid.points
QMI2D_ANCHOR_CELL = q2d_grid.anchor
qmi2d_cell_table = describe_qmi2d_grid(q2d_grid)
plot_qmi2d_grid(q2d_grid)
```

## Semente, modelos e sessão QMI2D

Nos pontos representativos das células donas, o campo inicial reproduz exatamente a onda-S
do ajuste QMI 1D em cada carga (verificado por `assert_allclose`). O NLL inicial não é o do
QMI 1D porque o campo é constante dentro de cada célula, enquanto a QMI 1D varia ao longo
de `s_high` pelo termo trocado `A(s_high)`.


```python
def points_sample(s_low, s_high):
    """Invariantes em (s13, s23) = (s_low, s_high), com s12 pela soma dos invariantes."""
    s_low, s_high = jnp.asarray(s_low, dtype=jnp.float64), jnp.asarray(s_high, dtype=jnp.float64)
    total = MOTHER_MASS ** 2 + sum(m ** 2 for m in DAUGHTER_MASSES)
    return {"s12": total - s_low - s_high, "s13": s_low, "s23": s_high}


def component_sum(model, names, fit_values, sample):
    """`sample`: dict com s12, s13, s23. Soma coerente coeficiente x dinâmica dos componentes `names` (sem normalização individual)."""
    total = jnp.zeros(jnp.shape(sample["s13"]), dtype=jnp.complex128)
    for component in model.amplitude_model.components:
        if component.name in names:
            total = total + (component.coefficient.value(fit_values)
                             * component.function(sample, fit_values))
    return np.asarray(total)


@dataclass(frozen=True)
class CPCellPolar:
    """|z| ou arg(z), com z = (x + q dx) + i (y + q dy), resolvido pela API de dinâmica.

    O QMI2D é parametrizado em magnitude/fase; este helper mantém os parâmetros
    cartesianos x, y, dx, dy por célula (como os bins da QMI 1D) e é diferenciável em JAX
    fora de z = 0.
    """
    x: Parameter
    y: Parameter
    dx: Parameter
    dy: Parameter
    charge: int
    part: str  # "mag" ou "phase"

    def resolve(self, values=None):
        re = self.x.resolve(values) + self.charge * self.dx.resolve(values)
        im = self.y.resolve(values) + self.charge * self.dy.resolve(values)
        return jnp.hypot(re, im) if self.part == "mag" else jnp.arctan2(im, re)


@dataclass(frozen=True)
class OppositeSignQMI2D:
    """Passa os dois invariantes pi+pi- (s13, s23) ao QMI2D dobrado.

    Dataclass para que os nós aninhados continuem visíveis ao coletor de parâmetros
    do DecayModel e ao resolvedor recursivo de valores (como no notebook 13 KKK).
    """
    field: QMI2D

    def __call__(self, data):
        return self.field({"s12": data["s13"], "s13": data["s23"]})


def build_qmi2d(grid, start_values, *, fixed_components=(), per_cell_cp=QMI2D_PER_CELL_CP):
    """Modelos B+/B- e sessão com a onda-S substituída pelo campo S_QMI2D na grade `grid`.

    Semente: onda-S do fit QMI 1D (`values`: coeficiente global x QMI simetrizada), por carga, no ponto
    representativo de cada célula dona, relativa à célula âncora da mesma carga. Os demais coeficientes
    partem de `start_values`; os de `fixed_components` ficam fixos nesses valores.
    """
    owner_cells = list(grid.free_cells)
    seed_points = points_sample(*np.asarray([grid.points[c] for c in owner_cells]).T)
    seed = {charge: component_sum(model, ("S_QMI",), values, seed_points)
            for charge, model in ((+1, plus_model), (-1, minus_model))}
    anchor_index = owner_cells.index(grid.anchor)
    global_seed = {charge: complex(s[anchor_index]) for charge, s in seed.items()}
    if any(not np.isfinite(v) or abs(v) < 1e-12 for v in global_seed.values()):
        raise ValueError("Semente nula/não finita na célula âncora")
    relative = {charge: s / global_seed[charge] for charge, s in seed.items()}
    if per_cell_cp:
        average, difference = (relative[+1] + relative[-1]) / 2, (relative[+1] - relative[-1]) / 2
    else:  # forma comum; a diferença entre cargas vai para o coeficiente global
        average, difference = relative[+1], np.zeros_like(relative[+1])

    cell_cp = {}
    for k, cell in enumerate(owner_cells):
        is_anchor = cell == grid.anchor
        seeds = (average[k].real, average[k].imag, difference[k].real, difference[k].imag)
        cell_cp[cell] = {
            key: Parameter.dynamics(
                f"S_QMI2D.cell_{cell[0]:02d}_{cell[1]:02d}.{key}",
                1.0 if (is_anchor and key == "x") else (0.0 if is_anchor else float(value)),
                owner="S_QMI2D",
                fixed=is_anchor or (key in ("dx", "dy") and not per_cell_cp),
                step=max(1e-4, 0.02 * abs(relative[+1][k]) + 0.02 * abs(relative[-1][k])),
            )
            for key, value in zip(("x", "y", "dx", "dy"), seeds)
        }

    def make_field(charge):
        def cells(part):
            return tuple(tuple(
                CPCellPolar(**cell_cp[grid.owner[(i, j)]], charge=charge, part=part)
                if (i, j) in grid.owner else 0.0
                for j in range(grid.shape[1])) for i in range(grid.shape[0]))

        return OppositeSignQMI2D(QMI2D(
            s12_edges=grid.edges, s13_edges=grid.edges,
            magnitudes=cells("mag"), phases=cells("phase"),
            active_mask=tuple(tuple(bool(v) for v in row) for row in grid.mask),
            folded=True, interpolation=QMI2D_INTERPOLATION,
        ))

    shared = make_shared_coefficients(start_values)
    for name in fixed_components:
        shared[name] = CPRealImag(*(
            Parameter.coefficient(f"{name}.{key}", float(start_values[f"{name}.{key}"]), owner=name, fixed=True)
            for key in ("x", "y", "dx", "dy")
        ))
    c_average = (global_seed[+1] + global_seed[-1]) / 2
    c_difference = (global_seed[+1] - global_seed[-1]) / 2
    shared["S_QMI2D"] = CPRealImag(*(
        Parameter.coefficient(f"S_QMI2D.{key}", float(value), owner="S_QMI2D", fixed=False)
        for key, value in zip(("x", "y", "dx", "dy"),
                              (c_average.real, c_average.imag, c_difference.real, c_difference.imag))
    ))
    dynamics = make_dynamics_parameters(start_values)

    def components(charge):
        coefficient = {name: cp.for_charge(charge) for name, cp in shared.items()}
        return [
            Resonance(
                name, RESONANCE_PAIR, coefficient[name],
                lineshape=LINESHAPES[name](), angular=ZemachP(),
                normalize_component=False,
                mass=dynamics[f"{name}.mass"], width=dynamics[f"{name}.width"], spin=SPINS[name],
                resonance_radius=4.0, parent_radius=4.0,
            )
            for name in LINESHAPES if name not in SWAVE_NAMES
        ] + [DalitzAmplitude("S_QMI2D", make_field(charge), coefficient["S_QMI2D"], normalize_component=False)]

    models = {charge: DecayModel(channel, components(charge), normalize_components=False, **NORMALIZATION_CONFIG)
              for charge, channel in ((+1, PLUS_CHANNEL), (-1, MINUS_CHANNEL))}
    session = make_data_fit_session(models[+1], models[-1])

    # Fechamento da semente: nos pontos representativos, o campo inicial reproduz a onda-S QMI 1D
    # exatamente nas células donas (as células fundidas herdam o valor da dona).
    if per_cell_cp:
        for charge in (+1, -1):
            np.testing.assert_allclose(component_sum(models[charge], ("S_QMI2D",), {}, seed_points),
                                       seed[charge], rtol=1e-8, atol=1e-10)
    n_free = sum(1 for p in session.parameters if not p.fixed)
    print(f"parâmetros: {len(session.parameters)} total, {n_free} livres"
          + (f"; fixos no QMI 1D: {', '.join(fixed_components)}" if fixed_components else ""))
    return SimpleNamespace(grid=grid, models=models, session=session, shared=shared,
                           cell_cp=cell_cp, seed_points=seed_points, seed=seed)


q2d = build_qmi2d(q2d_grid, values)
q2d_models, q2d_session, q2d_shared = q2d.models, q2d.session, q2d.shared
print("B+ componentes:", [c.name for c in q2d_models[+1].amplitude_model.components])

# Mesma grade e semente, com x, y, dx, dy do rho0_1450 fixos no resultado do QMI 1D.
q2d_rho_fixed = build_qmi2d(q2d_grid, values, fixed_components=("rho0_1450",))
```

## Frações de ajuste e ACP: referências × isobárico × QMI 1D × QMI 2D (livre e ρ(1450) fixo)

Integrais físicas no DP completo, sem eficiência/veto, `integral(f) = mean(weights * f)`, como nas
seções anteriores. A linha "S-wave" é a soma coerente `sigma0 + f0_980 + f0_1370` no isobárico
(interferência interna incluída), `S_QMI` no QMI 1D e o campo `S_QMI2D` nos dois QMI 2D.
`FF_CP` é ponderada pelas taxas R±; a "Ref. tabela global" dá `<FF>` sem definição explícita, e o
draft (Tab. 9) só FF_CP e ACP. As incertezas de ACP dos ajustes propagam a covariância completa
de cada ajuste (delta method; parâmetros fixos não contribuem); os ajustes usam os mesmos dados, então são correlacionados entre si.


```python
COMMON_COMPONENTS = ("rho0_770", "omega_782", "rho0_1450", "f2_1270", "rho0_3_1690", "chi_c0", "S-wave")
COMMON_LABELS = {**{k: v for k, v in REFERENCE_TABLE_NAMES.items() if k != "S_QMI"}, "S-wave": "S-wave"}


def fit_groups(fit):
    """Rótulo comum -> componentes do fit; a onda-S isobárica é a soma coerente dos três isóbaros."""
    return {name: (tuple(fit["swave"]) if name == "S-wave" else (name,)) for name in COMMON_COMPONENTS}


def group_rates(cache, component_names, groups, fit_values):
    """Taxas físicas |sum c_i F_i|^2 integradas por grupo (interferência interna incluída) e total."""
    matrix = cache.normalization_matrix(fit_values)
    coefficients = cache.coefficient_vector(fit_values)
    total = jnp.real(jnp.vdot(coefficients, matrix @ coefficients))
    rates = []
    for members in groups.values():
        idx = jnp.asarray([component_names.index(n) for n in members])
        c = coefficients[idx]
        rates.append(jnp.real(jnp.vdot(c, matrix[jnp.ix_(idx, idx)] @ c)))
    return jnp.stack(rates), total


def fit_observables(fit):
    """FF por carga, FF_CP e ACP (com erro pelo delta method) por grupo comum de componentes."""
    groups = fit_groups(fit)
    names = [c.name for c in fit["models"][+1].amplitude_model.components]
    # Integrais no DP completo, sem eficiência/veto (mesma convenção das FF físicas acima).
    caches = {q: model.prepare_cache(model.normalization_sample.take(jnp.arange(1)))
              for q, model in fit["models"].items()}

    def rates(fit_values):
        return {q: group_rates(cache, names, groups, fit_values) for q, cache in caches.items()}

    def acp_percent(fit_values):
        r = rates(fit_values)
        return 100.0 * (r[-1][0] - r[+1][0]) / (r[-1][0] + r[+1][0])

    r = rates(fit["values"])
    acp_errors = delta_method_errors(acp_percent, fit["values"], fit["free_names"], fit["result"].covariance)
    return pd.DataFrame({
        "FF B+ [%]": np.asarray(100 * r[+1][0] / r[+1][1]),
        "FF B- [%]": np.asarray(100 * r[-1][0] / r[-1][1]),
        "FF_CP [%]": np.asarray(100 * (r[+1][0] + r[-1][0]) / (r[+1][1] + r[-1][1])),
        "ACP [%]": np.asarray(acp_percent(fit["values"])),
        "sigma(ACP) [%]": np.asarray(acp_errors),
    }, index=list(COMMON_COMPONENTS))


observables = {label: fit_observables(fit) for label, fit in FITS.items()}

# Referências: tabela de coeficientes globais (NLL 967471) e Tabela 9 do draft (QMI 212-112).
_global_ref = GLOBAL_REFERENCE.rename(index={"S_QMI": "S-wave"})
observables["Ref. tabela global"] = pd.DataFrame({
    "FF B+ [%]": _global_ref.ff_plus, "FF B- [%]": _global_ref.ff_minus,
    "FF_CP [%]": _global_ref.ff_average, "ACP [%]": _global_ref.acp, "sigma(ACP) [%]": _global_ref.acp_error,
}).reindex(COMMON_COMPONENTS)
observables["Draft Tab. 9"] = pd.DataFrame({
    "FF_CP [%]": {k: PAPER_TABLE9_QMI[COMMON_LABELS[k]]["ff_percent"] for k in COMMON_COMPONENTS},
    "ACP [%]": {k: PAPER_TABLE9_QMI[COMMON_LABELS[k]]["acp_percent"] for k in COMMON_COMPONENTS},
    "sigma(ACP) [%]": {k: PAPER_TABLE9_QMI[COMMON_LABELS[k]]["acp_err_percent"] for k in COMMON_COMPONENTS},
}).reindex(COMMON_COMPONENTS)

SOURCES = ("Ref. tabela global", "Draft Tab. 9", *FITS)
three_fit_observables = pd.concat({s: observables[s] for s in SOURCES}, axis=1).swaplevel(axis=1)
three_fit_observables = three_fit_observables[
    [(q, s) for q in ("FF B+ [%]", "FF B- [%]", "FF_CP [%]", "ACP [%]", "sigma(ACP) [%]")
     for s in SOURCES if (q, s) in three_fit_observables.columns]
]
display(three_fit_observables.round(2))
display(pd.DataFrame({s: observables[s][["FF B+ [%]", "FF B- [%]", "FF_CP [%]"]].sum()
                      for s in FITS}).T.rename(columns=lambda c: "soma " + c))

SOURCE_STYLE = {
    "Ref. tabela global": dict(marker="s", color="tab:red"),
    "Draft Tab. 9": dict(marker="D", color="tab:purple"),
    "Isobárico": dict(marker="^", color="tab:green"),
    "QMI 1D": dict(marker="o", color="tab:blue"),
    "QMI 2D": dict(marker="v", color="tab:orange"),
    "QMI 2D ρ(1450) fixo": dict(marker="P", color="tab:brown"),
}
xpos = np.arange(len(COMMON_COMPONENTS))
fig, axes = plt.subplots(2, 2, figsize=(17, 10), constrained_layout=True)
for ax, quantity in zip(axes.flat, ("FF B+ [%]", "FF B- [%]", "FF_CP [%]", "ACP [%]")):
    for k, source in enumerate(SOURCES):
        frame = observables[source]
        if quantity not in frame:
            continue
        offset = (k - (len(SOURCES) - 1) / 2) * 0.12
        yerr = frame["sigma(ACP) [%]"] if quantity == "ACP [%]" else None
        ax.errorbar(xpos + offset, frame[quantity], yerr=yerr, fmt=SOURCE_STYLE[source]["marker"],
                    color=SOURCE_STYLE[source]["color"], ms=6, capsize=2, label=source)
    ax.set_title(quantity)
    ax.set_xticks(xpos, [COMMON_LABELS[k] for k in COMMON_COMPONENTS], rotation=45, ha="right")
    if quantity.startswith("FF"):
        ax.set_yscale("log")
    else:
        ax.axhline(0, color="grey", lw=0.8)
    ax.grid(alpha=0.3)
axes[0, 0].legend(fontsize=9)
plt.show()
```

```python
ISOBAR_COMMON = [k for k in COMMON_COMPONENTS if k != "S-wave"]


def fit_coefficient(fit, name, charge, fit_values=None):
    model = fit["models"][charge]
    component = next(c for c in model.amplitude_model.components if c.name == name)
    return complex(component.coefficient.value(fit["values"] if fit_values is None else fit_values))


coefficient_rows = []
for name in ISOBAR_COMMON:
    for charge, suffix in ((+1, "plus"), (-1, "minus")):
        row = {"component": name, "charge": suffix}
        for label, fit in FITS.items():
            coef, rho = fit_coefficient(fit, name, charge), fit_coefficient(fit, "rho0_770", charge)
            row[f"|c| {label}"] = abs(coef)
            row[f"fase {label}"] = float(_wrap_deg(np.degrees(np.angle(coef / rho))))
        row["fase Ref. tabela global"] = GLOBAL_REFERENCE.loc[name, f"phase_{suffix}"]
        row["fase Draft Tab. 23"] = PAPER_TABLE23_QMI[COMMON_LABELS[name]][f"delta_{suffix}"]
        coefficient_rows.append(row)
three_fit_coefficients = pd.DataFrame(coefficient_rows).set_index(["component", "charge"])
display(three_fit_coefficients.round(3))

xpos = np.arange(len(ISOBAR_COMMON))
fig, axes = plt.subplots(2, 2, figsize=(16, 9), constrained_layout=True)
for col, suffix in enumerate(("plus", "minus")):
    frame = three_fit_coefficients.xs(suffix, level="charge").loc[ISOBAR_COMMON]
    for k, label in enumerate(FITS):
        style = SOURCE_STYLE[label]
        axes[0, col].plot(xpos + (k - (len(FITS) - 1) / 2) * 0.15, frame[f"|c| {label}"], style["marker"],
                          color=style["color"], label=label)
    phase_sources = ("Ref. tabela global", "Draft Tab. 23", *FITS)
    for k, source in enumerate(phase_sources):
        style = SOURCE_STYLE.get(source, SOURCE_STYLE["Draft Tab. 9"])
        axes[1, col].plot(xpos + (k - (len(phase_sources) - 1) / 2) * 0.12, frame[f"fase {source}"], style["marker"],
                          color=style["color"], label=source)
    axes[0, col].set(yscale="log", ylabel="|coeficiente| (convenção deste notebook)",
                     title=f"B{'+' if suffix == 'plus' else '-'}: magnitudes (só os ajustes)")
    axes[1, col].set(ylabel="fase relativa ao rho [graus]", ylim=(-180, 180))
    for ax in axes[:, col]:
        ax.set_xticks(xpos, [COMMON_LABELS[k] for k in ISOBAR_COMMON], rotation=30, ha="right")
        ax.grid(alpha=0.3)
        ax.legend(fontsize=8)
plt.show()
```

## Onda-S no Dalitz dobrado

`|A_S|²` e fase (relativa ao rho da mesma carga) em `(s_low, s_high)` para a onda-S isobárica,
a `S_QMI` 1D simetrizada e o campo `S_QMI2D` dos dois ajustes 2D, com escala de cor comum (todos em unidades do rho).
As bandas brancas são o veto D0; as linhas finas no QMI 2D marcam as células.
A terceira linha compara as cargas: na intensidade, a assimetria local
`(|A−|² − |A+|²)/(|A−|² + |A+|²)` em [−1, 1] (mesma convenção de sinal da ACP integrada); na fase, `fase− − fase+` reduzida a ±180°, em mapa cíclico (0° claro, ±180° escuro) (as fases de cada carga são relativas ao rho da mesma carga).
A onda-S 1D tem a forma `A(s_low) + A(s_high)`; o campo 2D não tem essa restrição.


```python
from matplotlib.colors import LogNorm

# Grade regular no Dalitz dobrado (s_low, s_high); pontos fora do domínio físico ou no veto D0 -> NaN.
_map_low = np.linspace(_q2d_edges[0], 14.0, 300)
_map_high = np.linspace(_q2d_edges[0], _q2d_edges[-1], 400)
_mlow, _mhigh = np.meshgrid(_map_low, _map_high, indexing="ij")
_lim_lo, _lim_hi = map(np.asarray, dalitz_s13_limits(jnp.asarray(_mlow.ravel()), mother_mass=MOTHER_MASS,
                                                     masses=DAUGHTER_MASSES))
_in_veto = lambda s: (s > D0_VETO_MIN_GEV ** 2) & (s < D0_VETO_MAX_GEV ** 2)
_map_valid = ((_mhigh.ravel() >= np.maximum(_lim_lo, _mlow.ravel())) & (_mhigh.ravel() <= _lim_hi)
              & ~_in_veto(_mlow.ravel()) & ~_in_veto(_mhigh.ravel()))
_map_points = points_sample(_mlow.ravel()[_map_valid], _mhigh.ravel()[_map_valid])


def swave_amplitude(fit, charge, sample):
    """Onda-S (ou campo 2D) de um fit, com fase relativa ao coeficiente do rho da mesma carga."""
    amplitude = component_sum(fit["models"][charge], fit["swave"], fit["values"], sample)
    return amplitude * np.exp(-1j * np.angle(fit_coefficient(fit, "rho0_770", charge)))


swave_maps = {}
for label, fit in FITS.items():
    for charge in (+1, -1):
        full = np.full(_mlow.size, np.nan, dtype=complex)
        full[_map_valid] = swave_amplitude(fit, charge, _map_points)
        swave_maps[(label, charge)] = full.reshape(_mlow.shape)

_intensity_max = max(np.nanmax(np.abs(m) ** 2) for m in swave_maps.values())
# B- vs B+ por fit: assimetria local (|A-|^2 - |A+|^2) / (|A-|^2 + |A+|^2) e fase (fase- - fase+) em [-180, 180).
swave_differences = {
    label: {
        "intensity": ((np.abs(swave_maps[(label, -1)]) ** 2 - np.abs(swave_maps[(label, +1)]) ** 2)
                      / (np.abs(swave_maps[(label, -1)]) ** 2 + np.abs(swave_maps[(label, +1)]) ** 2)),
        "phase": _wrap_deg(np.degrees(np.angle(swave_maps[(label, -1)]) - np.angle(swave_maps[(label, +1)]))),
    }
    for label in FITS
}

for quantity in ("intensity", "phase"):
    fig, axes = plt.subplots(3, len(FITS), figsize=(5.6 * len(FITS), 13), sharex=True, sharey=True, constrained_layout=True)
    for row, charge in enumerate((+1, -1, None)):
        for col, label in enumerate(FITS):
            ax = axes[row, col]
            if charge is None:
                values_2d = swave_differences[label][quantity]
                if quantity == "intensity":
                    diff_mesh = ax.pcolormesh(_map_low, _map_high, values_2d.T, shading="auto",
                                              vmin=-1, vmax=1, cmap="RdBu_r")
                else:
                    diff_mesh = ax.pcolormesh(_map_low, _map_high, values_2d.T, shading="auto",
                                              vmin=-180, vmax=180, cmap="twilight_shifted")  # cíclico: 0 claro, ±180 escuro
                title = f"{'A_CP local' if quantity == 'intensity' else 'B- − B+'} — {label}"
            else:
                amp = swave_maps[(label, charge)]
                if quantity == "intensity":
                    mesh = ax.pcolormesh(_map_low, _map_high, (np.abs(amp) ** 2).T, shading="auto",
                                         norm=LogNorm(vmin=_intensity_max * 1e-4, vmax=_intensity_max), cmap="viridis")
                else:
                    mesh = ax.pcolormesh(_map_low, _map_high, np.degrees(np.angle(amp)).T, shading="auto",
                                         vmin=-180, vmax=180, cmap="twilight")
                title = f"B{'+' if charge > 0 else '-'} — {label}"
            draw_q2d_boundary(ax)
            if label.startswith("QMI 2D"):
                for edge in _q2d_edges[1:-1]:
                    ax.axvline(edge, color="grey" if charge is None else "white", lw=0.3, alpha=0.5)
                    ax.axhline(edge, color="grey" if charge is None else "white", lw=0.3, alpha=0.5)
            ax.set(xlim=(_map_low[0], _map_low[-1]), ylim=(_map_high[0], _map_high[-1]), title=title)
            if row == 2:
                ax.set_xlabel(r"$s_\mathrm{low}$ [GeV$^2$]")
            if col == 0:
                ax.set_ylabel(r"$s_\mathrm{high}$ [GeV$^2$]")
    fig.colorbar(mesh, ax=axes[:2], label=(r"$|A_S|^2$ (unidades do rho, escala comum)" if quantity == "intensity"
                                           else "fase relativa ao rho [graus]"))
    fig.colorbar(diff_mesh, ax=axes[2], label=(r"$(|A_S^-|^2 - |A_S^+|^2)\,/\,(|A_S^-|^2 + |A_S^+|^2)$" if quantity == "intensity"
                                               else r"fase$^-$ − fase$^+$ [graus]"))
    fig.suptitle("Onda-S isobárica (sigma+f0(980)+f0(1370)), S_QMI 1D e campo S_QMI2D (livre e com rho(1450) fixo)"
                 + (": intensidade" if quantity == "intensity" else ": fase"))
    plt.show()
```

## Goodness-of-fit dos quatro ajustes

Mesmo chi2 binado de Pearson das seções anteriores (Square-Dalitz dobrado e DP convencional dobrado,
25×25), com as cotas de Williams `dof_min`/`dof_max` (`docs/goodness_of_fit.md`). Com ~300 parâmetros
livres, `dof_min` dos QMI 2D é bem menor; compare os p-values pelas duas cotas, não só pelo chi2.


```python
# Mesmo chi2 binado das seções anteriores (Square-Dalitz dobrado e DP convencional dobrado, 25x25),
# agora para os quatro fits; o QMI 1D reutiliza gof_square_dalitz/gof_dp já calculados acima.
gof_all = {"QMI 1D": {"Square-Dalitz": gof_square_dalitz, "DP dobrado": gof_dp}}
for label in [name for name in FITS if name != "QMI 1D"]:
    fit = FITS[label]
    gof_all[label] = {
        "Square-Dalitz": fit["session"].goodness_of_fit_chi2(
            fit["result"], square_dalitz=True, folded=True, mother_mass=MOTHER_MASS,
            masses=DAUGHTER_MASSES, pair=SQDP_PAIR, bins=25, projection_size=1_000_000),
        "DP dobrado": fit["session"].goodness_of_fit_chi2(
            fit["result"], x="s13", y="s23", folded=True, bins=25, projection_size=1_000_000),
    }

gof_rows = []
for label in FITS:
    for plane, results in gof_all[label].items():
        for charge, res in results.items():
            gof_rows.append({"fit": label, "plano": plane, "carga": charge, "chi2": res.chi2,
                             "n_bins": res.n_bins, "n_livres": res.n_free_parameters,
                             "dof_min": res.dof_min, "dof_max": res.dof_max,
                             "p_min": res.p_value_min, "p_max": res.p_value_max})
gof_summary = pd.DataFrame(gof_rows).set_index(["plano", "carga", "fit"]).sort_index()
display(gof_summary)

fig, axes = plt.subplots(2, len(FITS), figsize=(5.6 * len(FITS), 9), sharex=True, sharey=True, constrained_layout=True)
_pull_max = max(np.nanmax(np.abs(res.pulls)) for label in FITS
                for res in gof_all[label]["Square-Dalitz"].values())
for col, label in enumerate(FITS):
    for row, (charge, res) in enumerate(gof_all[label]["Square-Dalitz"].items()):
        mp_edges, tp_edges = res.edges
        mesh = axes[row, col].pcolormesh(mp_edges, tp_edges, res.pulls.T, cmap="RdBu_r",
                                         vmin=-_pull_max, vmax=_pull_max, shading="flat")
        axes[row, col].set(title=f"B{'+' if charge == 'plus' else '-'} — {label}: chi2={res.chi2:.0f}",
                           xlabel="m'" if row == 1 else None, ylabel="theta'" if col == 0 else None)
fig.colorbar(mesh, ax=axes, label=r"pull $(o-e)/\sqrt{e}$")
plt.show()
```

### Distribuição 1D dos pulls por modelo

Pulls `(o − e)/√e` dos bins não vazios do chi2 acima, com um ajuste gaussiano (ML não binado) por
modelo; a curva tracejada é N(0, 1). O gráfico usa o DP convencional dobrado (como na parte 2) e a
tabela traz os dois planos. Bins com poucos eventos esperados não são gaussianos, então μ e σ servem
para comparar os modelos entre si, não como teste absoluto.


```python
from scipy.stats import norm

# Distribuição 1D dos pulls dos bins não vazios (mesmos chi2 acima), com ajuste gaussiano de ML
# não binado por modelo. Gráfico: DP convencional dobrado, como na parte 2; tabela: os dois planos.
pull_fit_rows = []
for plane in ("DP dobrado", "Square-Dalitz"):
    for label in FITS:
        for charge, res in gof_all[label][plane].items():
            pulls_1d = np.asarray(res.pulls).ravel()
            pulls_1d = pulls_1d[np.isfinite(pulls_1d)]
            mu, sigma = norm.fit(pulls_1d)
            pull_fit_rows.append({"plano": plane, "carga": charge, "fit": label, "n_bins": pulls_1d.size,
                                  "mu": mu, "erro mu": sigma / np.sqrt(pulls_1d.size),
                                  "sigma": sigma, "erro sigma": sigma / np.sqrt(2 * pulls_1d.size)})
pull_fit_summary = pd.DataFrame(pull_fit_rows).set_index(["plano", "carga", "fit"]).sort_index()
display(pull_fit_summary.round(3))

_pull_range = max(np.nanmax(np.abs(res.pulls)) for label in FITS for res in gof_all[label]["DP dobrado"].values())
_pull_bins = np.linspace(-_pull_range, _pull_range, 31)
_pull_grid = np.linspace(-_pull_range, _pull_range, 400)
_pull_width = _pull_bins[1] - _pull_bins[0]
fig, axes = plt.subplots(1, 2, figsize=(15, 5), sharey=True, constrained_layout=True)
for ax, charge in zip(axes, ("plus", "minus")):
    for label in FITS:
        row = pull_fit_summary.loc[("DP dobrado", charge, label)]
        pulls_1d = np.asarray(gof_all[label]["DP dobrado"][charge].pulls).ravel()
        pulls_1d = pulls_1d[np.isfinite(pulls_1d)]
        color = SOURCE_STYLE[label]["color"]
        ax.hist(pulls_1d, bins=_pull_bins, histtype="step", color=color, lw=1.3)
        ax.plot(_pull_grid, row.n_bins * _pull_width * norm.pdf(_pull_grid, row.mu, row.sigma), color=color, lw=1.8,
                label=rf"{label}: $\mu={row.mu:.2f}\pm{row['erro mu']:.2f}$, $\sigma={row.sigma:.2f}\pm{row['erro sigma']:.2f}$")
    n_ref = pull_fit_summary.loc[("DP dobrado", charge), "n_bins"].iloc[0]
    ax.plot(_pull_grid, n_ref * _pull_width * norm.pdf(_pull_grid), "k--", lw=1, label=r"$\mathcal{N}(0,1)$")
    ax.set(xlabel=r"pull $(o-e)/\sqrt{e}$", title=f"B{'+' if charge == 'plus' else '-'}: pulls do DP dobrado (25×25)")
    ax.legend(fontsize=8, loc="upper left")
axes[0].set_ylabel("bins")
plt.show()
```

## Limites da comparação QMI2D

- A grade (`QMI2D_EDGES_GEV2`), o limiar de fusão (`QMI2D_MIN_EVENTS`) e a âncora são escolhas
  explícitas, não validadas por toys. Antes de interpretar: repetir com outra âncora,
  `QMI2D_PER_CELL_CP=False` e mais starts, e verificar a estabilidade de FF/ACP.
- Uma grade mais grossa em 1.0–1.74 GeV (só 1.2 e 1.5 GeV) deu NLL 114.5 pior e FF do ρ(1450) em B+
  de 0.17% (ACP +90%): o ajuste compensava a onda-S mal descrita zerando o ρ(1450). A grade atual
  resolve isso, mas mais células não fazem o resultado convergir para o QMI 1D: o campo não é spin 0
  e, quanto mais fino, mais consegue imitar a onda P do ρ(1450).
- O campo pode absorver componentes largas (em especial `rho0_1450`): FF/ACP desses componentes no
  QMI 2D livre não são comparáveis diretamente com as dos modelos com onda-S 1D. Compare com o ajuste
  de ρ(1450) fixo e com o ΔNLL entre eles.
- `valid=True` e covariância precisa não garantem ausência de direções quase degeneradas; inspecionar
  as correlações do bloco `S_QMI2D` × `rho0_1450` antes de citar incertezas.
